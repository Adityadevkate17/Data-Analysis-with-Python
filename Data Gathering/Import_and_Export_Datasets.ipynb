# Export Datasets

Pandas Export

    . to csv
    . to excel
    . to html
    . to json
    . to sql

```python
import pandas as pd
df = pd.DataFrame()
```

## to_csv

```python
df= pd.read_csv('deliveries.csv')
df
```

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
```

```python
# Add this dataset in folder
temp_df.to_csv('batsman_runs.csv',index=False)
```

```python
df.pivot_table(index='batsman',columns='bowling_team',values='batsman_runs',aggfunc='sum').to_csv('batsman_vs_team',index=True)
```

## to_excel

```python
temp_df = df.groupby('batsman')['batsman_runs'].sum().reset_index()
```

```python
temp_df
```

```python
# csv to excel
temp_df.to_excel('batsman_runs.xlsx')
```

```python
# change the sheet name
temp_df.to_excel('output.xlsx',sheet_name='batsman_runs')
```

```python
temp_df2 = df.pivot_table(index='batsman',columns='bowling_team', values='batsman_runs',aggfunc='sum')
```

```python
with pd.ExcelWriter('outputt.xlsx')as writer:
    temp_df.to_excel(writer,sheet_name='sheet_name_1')
    temp_df2.to_excel(writer,sheet_name='sheet_name_2')
```

## to_html

```python
df
```

```python
df.query('batsman_runs == 6').pivot_table(index='over',columns='ball',values='batsman_runs',aggfunc='count').to_html('sixes_heatmap.html')
```

```python
df
```

```python
from sqlalchemy import create_engine

# Updated connection string to use the pymysql driver
engine = create_engine("mysql+pymysql://root:@localhost/ipl")
# {root} : {password}@{url}/{database}

# Export your DataFrame to the MySQL table
df.to_sql('ipl_delivery', con=engine, if_exists='append', index=False)
```
